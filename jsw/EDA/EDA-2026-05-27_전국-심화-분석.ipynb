# 전체 심화 분석 EDA

기존 날씨·공간·통합 EDA를 직접 읽어 누락된 분석 축을 점검하고, 기존 `fire_spatial_features.csv`에 집계된 300m 면적 비율만으로는 확인하기 어려웠던 원천 토지피복도 기반 맥락을 보완한다.

이번 노트북의 핵심은 다음 네 가지다.

- 기존 EDA가 다룬 것과 놓친 것을 노트북 파일 기준으로 점검한다.
- `산불발생_토지피복도` 원천 GPKG에서 산불 지점 포함 피복, 산림·시가화·농업지역 최단거리, 산림 8방위 분포를 직접 산출한 캐시를 사용한다.
- 발생원인명, 도심/농지/산림 내부·경계 맥락, 산림 방향성과 진화/피해 결과를 연결한다.
- 기존 캐시에서 `total_lc_area_norm`만 채워지고 실제 토지피복 면적이 0으로 남은 원천 매칭 누락 문제를 별도 데이터 품질 이슈로 정리한다.

## 1. 기존 EDA 파일 직접 점검


```python
NOTEBOOKS = [
    "EDA-2026-05-26_전국-공간.ipynb",
    "EDA-2026-05-26_강원도-공간.ipynb",
    "EDA-2026-05-27_전국-날씨-공간-통합.ipynb",
    "EDA-2026-05-27_강원도-날씨-공간-통합.ipynb",
]

keywords = {
    "토지피복 대분류": ["산림지역_ratio", "농업지역_ratio", "시가화건조지역_ratio", "초지_ratio", "나지_ratio"],
    "산림 수종": ["침엽수림_ratio", "활엽수림_ratio", "혼효림_ratio"],
    "발생원인": ["발생원인명", "원인"],
    "산림 방향/거리": ["forest_dominant", "nearest_forest", "8방위", "동서남북", "경계"],
    "도심/시가화": ["시가화건조지역", "도심", "주거"],
    "접근성": ["nearest_station", "nearest_water", "nearest_road", "access_burden"],
}

audit_rows = []
heading_rows = []
for nb_name in NOTEBOOKS:
    nb_path = EDA_DIR / nb_name
    nb = json.loads(nb_path.read_text(encoding="utf-8"))
    full_text = "\n".join("".join(cell.get("source", [])) for cell in nb["cells"])
    row = {"노트북": nb_name, "셀 수": len(nb["cells"])}
    for label, kws in keywords.items():
        row[label] = sum(full_text.count(k) for k in kws)
    audit_rows.append(row)
    for idx, cell in enumerate(nb["cells"], start=1):
        if cell.get("cell_type") != "markdown":
            continue
        for line in "".join(cell.get("source", [])).splitlines():
            if line.strip().startswith("#"):
                heading_rows.append({"노트북": nb_name, "셀": idx, "제목": line.strip()})

audit_df = pd.DataFrame(audit_rows)
headings_df = pd.DataFrame(heading_rows)
display(audit_df)
display(headings_df.head(60))
```

### 1-1. 기존 EDA 점검 결과

기존 노트북은 기상, 지형, 접근성, 강원 영동/영서 상호작용을 충분히 다뤘다. 반면 다음 항목은 명시적으로 부족하다.

- `발생원인명`은 구조 확인 수준에 머물렀고, 토지피복 맥락과 연결되지 않았다.
- 토지피복도는 300m 반경 **면적 비율** 중심으로 집계됐고, 발생점이 실제 어느 피복 위에 있는지와 산림이 어느 방향/거리로 놓였는지는 분석하지 않았다.
- `시가화건조지역`, `농업지역`, `초지`, `나지`는 기초 평균 비교만 있었고 도심 인접형/농지 소각형/산림 경계형 같은 상황 분류로 확장되지 않았다.
- `nearest_road_dist_km`는 일반 도로가 아니라 임도망도 기준이므로, 도심 접근성이나 도로변 발화 해석에는 한계가 있다.
- 수종 세분류의 이상하게 낮은 비율은 원천 `자료출처(SG03/SG05)` 차이를 따로 검증해야 한다.


## 2. 원천 토지피복 심화 피처 로드 및 품질 확인


```python
ctx = pd.read_csv(CONTEXT_PATH, encoding="utf-8-sig", low_memory=False)

extra_cols = [
    "fire_id",
    "nearest_station_dist_km",
    "nearest_water_dist_km",
    "nearest_facility_dist_km",
    "nearest_road_dist_km",
    "nearest_trail_dist_km",
    "nearest_facility_remote",
    "산림지역_ratio",
    "시가화건조지역_ratio",
    "농업지역_ratio",
    "초지_ratio",
    "나지_ratio",
    "total_lc_area",
    "total_lc_area_norm",
]
base = pd.read_csv(BASE_FEATURE_PATH, usecols=extra_cols)
df = ctx.merge(base, on="fire_id", how="left", suffixes=("", "_old"))

df["_suppression_hours"] = pd.to_numeric(df["진화소요시간(HH)"], errors="coerce")
df["_suppression_valid"] = df["_suppression_hours"].gt(0) & df["_suppression_hours"].notna()
df["_large_fire_12h"] = np.where(df["_suppression_valid"], df["_suppression_hours"].ge(12), np.nan)
df["_damage_ha"] = pd.to_numeric(df["피해면적(ha)"], errors="coerce")
df["_large_damage_10ha"] = df["_damage_ha"].ge(10)
df["landcover_context_type"] = df["landcover_context_type"].fillna("원천매칭없음")
df["point_landcover_major"] = df["point_landcover_major"].fillna("원천매칭없음")
df["is_mountain_parcel"] = df["발생지역번지"].astype(str).str.contains("산", na=False)

yeongdong = {"강릉", "속초", "동해", "삼척", "태백", "고성", "양양", "강릉시", "속초시", "동해시", "삼척시", "태백시", "고성군", "양양군"}
df["gangwon_region"] = np.select(
    [df["is_gangwon"] & df["발생지역시군구명"].astype(str).isin(yeongdong), df["is_gangwon"]],
    ["강원 영동", "강원 영서"],
    default="타 지역",
)

def cause_group(value):
    text = str(value)
    if text in ["0", "미상", "원인미상", "조사중", "nan", "None", ""]:
        return "미상/미기록"
    if any(k in text for k in ["논", "밭", "두렁", "농산", "쓰레기", "소각", "폐기물"]):
        return "소각/농업활동"
    if any(k in text for k in ["입산", "등산", "성묘", "산나물", "약초"]):
        return "입산/산림활동"
    if any(k in text for k in ["담뱃", "담배", "차량", "도로변"]):
        return "도로/담뱃불"
    if any(k in text for k in ["주택", "가옥", "건축", "보일러", "아궁이"]):
        return "주거/시설 비화"
    if any(k in text for k in ["용접", "공사", "작업"]):
        return "공사/작업"
    if any(k in text for k in ["방화", "불장난"]):
        return "방화/불장난"
    if "낙뢰" in text:
        return "자연/낙뢰"
    return "기타"

df["cause_group"] = df["발생원인명"].map(cause_group)

print(f"분석 행 수: {len(df):,}")
print(f"진화소요시간 유효 표본: {df['_suppression_valid'].sum():,}")
print(f"원천 토지피복 매칭 없음: {(df['landcover_context_type'] == '원천매칭없음').sum():,}")
display(df.head(3))
```

```python
missing_raw = df["landcover_context_type"].eq("원천매칭없음")
quality_summary = pd.DataFrame({
    "전체 N": [len(df)],
    "원천매칭없음 N": [missing_raw.sum()],
    "원천매칭없음 비율(%)": [missing_raw.mean() * 100],
    "기존 캐시 total_lc_area=0 N": [(df["total_lc_area"].fillna(-1).eq(0)).sum()],
    "기존 캐시 ratio 전부 0이면서 norm만 존재 N": [
        ((df["total_lc_area"].fillna(-1).eq(0)) & (df["total_lc_area_norm"].fillna(0).gt(0))).sum()
    ],
})
display(quality_summary.round(3))

province_source = (
    df.groupby("발생지역시도명")
    .agg(
        N=("fire_id", "count"),
        원천매칭없음률=("landcover_context_type", lambda s: s.eq("원천매칭없음").mean() * 100),
        SG05평균비율=("sg05_area_ratio", "mean"),
        기존산림비율평균=("산림지역_ratio", "mean"),
        원천산림비율평균=("raw_forest_ratio", "mean"),
    )
    .sort_values("원천매칭없음률", ascending=False)
)
display(province_source.round(3))
```

### 2-1. 품질 확인 결과

원천 토지피복 GPKG에서 직접 매칭되는 산불은 22,250건이고, 2,786건은 원천 폴리곤이 매칭되지 않았다. 기존 `fire_spatial_features.csv`에서는 이 건들이 `total_lc_area_norm`은 300m 원 면적처럼 채워져 있지만 실제 `total_lc_area`와 모든 피복 비율은 0으로 남아 있었다. 따라서 기존 분석에서 `산림지역_ratio=0`으로 보인 일부 사례는 “비산림 도심 산불”이 아니라 **토지피복 원천 매칭 없음**일 수 있다.

또한 `SG05` 세분류 비율이 낮은 시도에서는 침엽수림·활엽수림·혼효림 비율이 낮게 보일 수 있다. 강원도는 `SG05` 평균 비율이 낮고 원천 매칭 없음도 있어, 이전 수종 세분류 분석을 핵심 피처로 쓰기 어렵다는 판단이 강화된다.


## 3. 발생점 피복과 300m 주변 맥락 유형


```python
def summarize_group(group_col):
    rows = []
    for key, g in df.groupby(group_col, dropna=False):
        valid = g[g["_suppression_valid"]]
        rows.append({
            group_col: key,
            "전체 N": len(g),
            "진화 유효 N": len(valid),
            "평균 진화시간(h)": valid["_suppression_hours"].mean(),
            "중앙 진화시간(h)": valid["_suppression_hours"].median(),
            "12h 이상 비율(%)": valid["_large_fire_12h"].mean() * 100,
            "10ha 이상 비율(%)": g["_large_damage_10ha"].mean() * 100,
            "평균 피해면적(ha)": g["_damage_ha"].mean(),
            "강원 비율(%)": g["is_gangwon"].mean() * 100,
            "산림 평균 비율": g["raw_forest_ratio"].mean(),
            "시가화 평균 비율": g["raw_built_ratio"].mean(),
            "농업 평균 비율": g["raw_agri_ratio"].mean(),
        })
    return pd.DataFrame(rows).sort_values("전체 N", ascending=False)

context_summary = summarize_group("landcover_context_type")
point_summary = summarize_group("point_landcover_major")
display(context_summary.round(4))
display(point_summary.round(4))
```

```python
order = context_summary["landcover_context_type"].tolist()
plot_df = (
    df.groupby(["gangwon_region", "landcover_context_type"], observed=False)
    .size()
    .reset_index(name="N")
)
plot_df["비율(%)"] = plot_df["N"] / plot_df.groupby("gangwon_region")["N"].transform("sum") * 100

rate_df = context_summary.copy()
fig, axes = plt.subplots(1, 2, figsize=(17, 6))
sns.barplot(data=plot_df, x="landcover_context_type", y="비율(%)", hue="gangwon_region", order=order, ax=axes[0])
axes[0].set_title("지역별 300m 토지피복 맥락 유형 분포")
axes[0].set_xlabel("")
axes[0].tick_params(axis="x", rotation=35)

sns.barplot(data=rate_df, x="landcover_context_type", y="12h 이상 비율(%)", order=order, ax=axes[1], color="#d36f45")
axes[1].set_title("맥락 유형별 12h 이상 진화 비율")
axes[1].set_xlabel("")
axes[1].tick_params(axis="x", rotation=35)
plt.tight_layout()
fig.savefig(TMP_PLOT_DIR / "01_context_distribution.png", dpi=160, bbox_inches="tight")
plt.show()
```

### 3-1. 발생점·주변 맥락 해석 및 플롯 이미지 분석

발생점 자체가 `산림지역`인 사례가 가장 많지만, `농업지역`, `시가화건조지역`, `초지`, `나지` 위에서 시작한 산불도 적지 않다. 이 때문에 300m 반경 토지피복은 “산림 수종”보다 **발생 위치가 산림 내부인지, 농지/도심과 맞닿은 경계인지**를 먼저 구분하는 피처가 더 설명력이 있다.

맥락 유형별로는 `산림 내부형`, `산림 우세형`, `산림 경계/혼재형`이 진화 유효 표본과 12h 이상 비율에서 상대적으로 높다. 반대로 `도심/시가화 인접형`은 건수는 많지만 평균 피해면적과 10ha 이상 비율이 낮아, 빠른 신고·접근성 또는 비연속 연료 구조가 작용했을 가능성이 있다.


## 4. 산림까지의 거리와 산림 8방위 분포


```python
forest_bins = [-0.1, 0, 30, 100, 300, 1e9]
forest_labels = ["발생점 포함", "0-30m", "30-100m", "100-300m", "300m내 산림 없음"]
df["forest_dist_band"] = pd.cut(
    pd.to_numeric(df["nearest_forest_dist_m"], errors="coerce").fillna(999999),
    bins=forest_bins,
    labels=forest_labels,
)
forest_dist_summary = summarize_group("forest_dist_band")
display(forest_dist_summary.round(4))

sector_order = ["N", "NE", "E", "SE", "S", "SW", "W", "NW"]
direction_df = df[df["forest_dominant_direction"].isin(sector_order)].copy()
direction_summary = (
    direction_df.groupby("forest_dominant_direction")
    .apply(lambda g: pd.Series({
        "전체 N": len(g),
        "진화 유효 N": g["_suppression_valid"].sum(),
        "평균 진화시간(h)": g.loc[g["_suppression_valid"], "_suppression_hours"].mean(),
        "12h 이상 비율(%)": g.loc[g["_suppression_valid"], "_large_fire_12h"].mean() * 100,
        "평균 피해면적(ha)": g["_damage_ha"].mean(),
        "강원 비율(%)": g["is_gangwon"].mean() * 100,
    }), include_groups=False)
    .reindex(sector_order)
)
display(direction_summary.round(4))

yd_direction_summary = (
    df[(df["gangwon_region"].eq("강원 영동")) & (df["forest_dominant_direction"].isin(sector_order))]
    .groupby("forest_dominant_direction")
    .apply(lambda g: pd.Series({
        "전체 N": len(g),
        "진화 유효 N": g["_suppression_valid"].sum(),
        "평균 진화시간(h)": g.loc[g["_suppression_valid"], "_suppression_hours"].mean(),
        "12h 이상 비율(%)": g.loc[g["_suppression_valid"], "_large_fire_12h"].mean() * 100,
        "평균 피해면적(ha)": g["_damage_ha"].mean(),
    }), include_groups=False)
    .reindex(sector_order)
)
display(yd_direction_summary.round(4))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
dir_share = (
    pd.crosstab(df["gangwon_region"], df["forest_dominant_direction"], normalize="index") * 100
).reindex(columns=sector_order + ["없음"], fill_value=0)
sns.heatmap(dir_share, annot=True, fmt=".1f", cmap="YlGnBu", ax=axes[0])
axes[0].set_title("지역별 산림 우세 방위 분포(%)")
axes[0].set_xlabel("산림 우세 방위")
axes[0].set_ylabel("")

sns.barplot(data=direction_summary.reset_index(), x="forest_dominant_direction", y="12h 이상 비율(%)", order=sector_order, ax=axes[1], color="#5a8f7b")
axes[1].set_title("산림 우세 방위별 12h 이상 진화 비율")
axes[1].set_xlabel("산림 우세 방위")
plt.tight_layout()
fig.savefig(TMP_PLOT_DIR / "02_forest_direction.png", dpi=160, bbox_inches="tight")
plt.show()
```

### 4-1. 산림 거리·방향 해석 및 플롯 이미지 분석

산불 발생점이 산림 폴리곤 안에 포함되는 경우가 가장 위험 신호가 강하다. 산림까지 거리가 30m 이상 멀어질수록 평균 피해와 12h 이상 비율이 급격히 낮아져, 단순 300m 반경 산림비율보다 **발생점과 산림의 접촉 여부**가 더 직접적인 피처가 될 수 있다.

산림 우세 방위는 전국 기준으로 어느 한 방향이 압도적이지 않지만, 강원 영동에서는 일부 방향에서 평균 피해면적이 크게 튄다. 다만 영동 방향별 표본은 작은 칸이 있으므로, 방위 단독 규칙보다 기존의 `사면방향 × 풍향 × 영동권` 상호작용과 결합하는 방식이 안전하다.

## 8. 발생일 악기상과 토지피복 맥락의 결합

기존 심화 분석은 원천 토지피복 맥락을 보완했지만, 그 맥락이 기상 피크 상황에서 어떻게 달라지는지는 아직 분리하지 않았다.
여기서는 발생일 날씨만 산불 ID 기준으로 결합하고, `평균 풍속 >= 3.0m/s` 또는 `최대 순간풍속 >= 7.0m/s` 또는 `최소 상대습도 <= 30%` 중 하나라도 만족하면 `악기상`으로 둔다.
결측 보간은 하지 않고, 진화소요시간 해석은 기존 노트북과 동일하게 0보다 큰 유효 표본만 사용한다.

```python
WEATHER_DAY_PATH = BASE_DIR / "data" / "날씨데이터" / "날씨데이터_일자별_전국산불.csv"
weather_day = pd.read_csv(WEATHER_DAY_PATH, encoding="utf-8-sig", low_memory=False)
weather_occ = (
    weather_day[weather_day["날짜구분"].eq("발생일")]
    .rename(columns={"산불아이디": "fire_id"})
    [[
        "fire_id",
        "평균 풍속",
        "최대 순간풍속",
        "최소 상대습도",
        "평균 상대습도",
        "평균 기온",
        "일강수량",
    ]]
)

weather_drop_cols = [
    "평균 풍속",
    "최대 순간풍속",
    "최소 상대습도",
    "평균 상대습도",
    "평균 기온",
    "일강수량",
    "weather_valid_day",
    "wind_peak_day",
    "dry_peak_day",
    "weather_peak_day",
    "weather_peak_label",
]
df = df.drop(columns=weather_drop_cols, errors="ignore").merge(weather_occ, on="fire_id", how="left").copy()

weather_valid = df[["평균 풍속", "최대 순간풍속", "최소 상대습도"]].notna().any(axis=1)
df["weather_valid_day"] = weather_valid
df["wind_peak_day"] = (df["평균 풍속"].ge(3.0) | df["최대 순간풍속"].ge(7.0)) & weather_valid
df["dry_peak_day"] = df["최소 상대습도"].le(30.0) & weather_valid
df["weather_peak_day"] = (df["wind_peak_day"] | df["dry_peak_day"]) & weather_valid
df["weather_peak_label"] = np.select(
    [~weather_valid, df["weather_peak_day"]],
    ["날씨결측", "악기상"],
    default="평시",
)
df["weather_peak_label"] = pd.Categorical(df["weather_peak_label"], ["평시", "악기상", "날씨결측"], ordered=True)

weather_quality_summary = pd.DataFrame({
    "전체 N": [len(df)],
    "발생일 날씨 유효 N": [df["weather_valid_day"].sum()],
    "악기상 N": [df["weather_peak_day"].sum()],
    "강풍 피크 N": [df["wind_peak_day"].sum()],
    "저습 피크 N": [df["dry_peak_day"].sum()],
    "진화 유효 N": [df["_suppression_valid"].sum()],
})
display(weather_quality_summary)

def summarize_weather_group(g):
    valid = g[g["_suppression_valid"]]
    damage = g["_damage_ha"].dropna()
    return pd.Series({
        "전체 N": len(g),
        "진화 유효 N": len(valid),
        "평균 진화시간(h)": valid["_suppression_hours"].mean(),
        "12h 이상 비율(%)": valid["_large_fire_12h"].mean() * 100 if len(valid) else np.nan,
        "10ha 이상 비율(%)": g["_large_damage_10ha"].mean() * 100,
        "평균 피해면적(ha)": damage.mean(),
        "중앙 피해면적(ha)": damage.median(),
        "90p 피해면적(ha)": damage.quantile(0.9),
        "최대 피해면적(ha)": damage.max(),
    })

context_weather_summary = (
    df.groupby(["landcover_context_type", "weather_peak_label"], observed=False, dropna=False)
    .apply(summarize_weather_group, include_groups=False)
    .reset_index()
)
display(
    context_weather_summary
    .sort_values(["weather_peak_label", "12h 이상 비율(%)"], ascending=[True, False])
    .round(4)
)
```

```python
context_order = [
    "산림 내부형",
    "산림 우세형",
    "산림 경계/혼재형",
    "농지/소각 인접형",
    "도심/시가화 인접형",
    "비산림 주변형",
    "원천매칭없음",
]
weather_order = ["평시", "악기상"]

heat = (
    context_weather_summary
    .pivot(index="landcover_context_type", columns="weather_peak_label", values="12h 이상 비율(%)")
    .reindex(context_order)[weather_order]
)
heat_n = (
    context_weather_summary
    .pivot(index="landcover_context_type", columns="weather_peak_label", values="진화 유효 N")
    .reindex(context_order)[weather_order]
)
heat_annot = heat.copy().astype(object)
for row in heat_annot.index:
    for col in heat_annot.columns:
        value = heat.loc[row, col]
        n_value = heat_n.loc[row, col]
        heat_annot.loc[row, col] = "" if pd.isna(value) else f"{value:.1f}%\nN={int(n_value)}"

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
sns.heatmap(
    heat,
    annot=heat_annot,
    fmt="",
    cmap="YlOrRd",
    linewidths=0.5,
    ax=axes[0],
    cbar_kws={"label": "12h 이상 비율(%)"},
)
axes[0].set_title("발생일 악기상 여부 × 토지피복 맥락")
axes[0].set_xlabel("발생일 기상")
axes[0].set_ylabel("")

weather_bar = context_weather_summary[context_weather_summary["weather_peak_label"].isin(weather_order)].copy()
weather_bar["landcover_context_type"] = pd.Categorical(weather_bar["landcover_context_type"], context_order, ordered=True)
sns.barplot(
    data=weather_bar.sort_values("landcover_context_type"),
    x="landcover_context_type",
    y="10ha 이상 비율(%)",
    hue="weather_peak_label",
    hue_order=weather_order,
    ax=axes[1],
)
axes[1].set_title("10ha 이상 피해 비율 비교")
axes[1].set_xlabel("")
axes[1].tick_params(axis="x", rotation=35)
axes[1].legend(title="발생일 기상")
plt.tight_layout()
fig.savefig(TMP_PLOT_DIR / "05_weather_landcover_context.png", dpi=160, bbox_inches="tight")
plt.show()
```

### 8-1. 발생일 악기상 결합 해석 및 플롯 이미지 분석

발생일 악기상 조건은 전체 25,036건 중 19,519건으로 넓게 잡히며, 이 조건에서 토지피복 맥락별 위험 차이가 더 뚜렷해졌다. 플롯을 직접 확인하면 평시 구간은 유효 진화 표본이 작아 색이 진하게 보이는 일부 칸(`비산림 주변형`, `원천매칭없음`)을 안정적 위험 신호로 읽기 어렵다.

`원천매칭없음`은 악기상에서 12h 이상 비율이 **7.03%**로 높지만 토지피복 의미가 결여된 품질 이슈 집단이다. 모델링에서는 위험 유형으로 해석하지 않고 `landcover_missing_flag` 같은 별도 결측/품질 플래그로 분리해야 한다.

## 9. 발생원인 × 토지피복 × 악기상 3중 조합

원인별 단순 평균은 `미상/미기록` 비중이 커서 직접적인 행동 가설로 이어지기 어렵다.
따라서 발생원인 그룹, 토지피복 맥락, 발생일 악기상을 함께 묶고, 전체 표본 40건 이상 및 진화 유효 표본 15건 이상인 조합만 별도로 확인한다.

```python
cause_weather_summary = (
    df.groupby(["cause_group", "landcover_context_type", "weather_peak_label"], observed=False, dropna=False)
    .apply(summarize_weather_group, include_groups=False)
    .reset_index()
)
cause_baseline = (
    df.groupby("cause_group")
    .apply(lambda g: pd.Series({
        "원인 기준 12h율(%)": g.loc[g["_suppression_valid"], "_large_fire_12h"].mean() * 100,
        "원인 N": len(g),
    }), include_groups=False)
    .reset_index()
)
cause_weather_summary = cause_weather_summary.merge(cause_baseline, on="cause_group", how="left")
cause_weather_summary["원인대비 12h 리프트"] = (
    cause_weather_summary["12h 이상 비율(%)"]
    / cause_weather_summary["원인 기준 12h율(%)"].replace(0, np.nan)
)

actionable_weather_risk = (
    cause_weather_summary[
        cause_weather_summary["weather_peak_label"].eq("악기상")
        & cause_weather_summary["전체 N"].ge(40)
        & cause_weather_summary["진화 유효 N"].ge(15)
        & ~cause_weather_summary["cause_group"].isin(["미상/미기록"])
        & ~cause_weather_summary["landcover_context_type"].eq("원천매칭없음")
    ]
    .sort_values(["12h 이상 비율(%)", "90p 피해면적(ha)"], ascending=[False, False])
    .reset_index(drop=True)
)
display(actionable_weather_risk.head(15).round(4))
```

```python
top_actionable_plot = actionable_weather_risk.head(12).copy()
top_actionable_plot["조합"] = top_actionable_plot["cause_group"] + "\n" + top_actionable_plot["landcover_context_type"]

fig, ax = plt.subplots(figsize=(11, 7))
sns.barplot(
    data=top_actionable_plot,
    y="조합",
    x="12h 이상 비율(%)",
    color="#b85b75",
    ax=ax,
)
x_max = top_actionable_plot["12h 이상 비율(%)"].max()
ax.set_xlim(0, x_max * 1.24)
for i, (_, row) in enumerate(top_actionable_plot.iterrows()):
    ax.text(
        row["12h 이상 비율(%)"] + x_max * 0.015,
        i,
        f"N={int(row['진화 유효 N'])}, 10ha={row['10ha 이상 비율(%)']:.1f}%",
        va="center",
        fontsize=9,
    )
ax.set_title("악기상 조건의 발생원인 × 토지피복 고위험 조합")
ax.set_xlabel("12h 이상 진화 비율(%)")
ax.set_ylabel("")
plt.tight_layout()
fig.savefig(TMP_PLOT_DIR / "06_cause_context_weather_peak.png", dpi=160, bbox_inches="tight")
plt.show()
```

### 9-1. 발생원인·토지피복·악기상 조합 해석 및 플롯 이미지 분석

플롯을 직접 확인하면 상위 조합은 대부분 산림 내부 또는 산림 우세 맥락에 몰려 있다. `방화/불장난 × 산림 내부형 × 악기상`은 12h 이상 비율 **5.00%**와 10ha 이상 비율 **5.00%**를 동시에 보여 강한 신호지만, 진화 유효 표본이 60건이라 고정 규칙보다는 후보 조합으로 다루는 편이 안전하다.

표본이 큰 조합에서는 `입산/산림활동 × 산림 내부형 × 악기상`이 진화 유효 **1,459건**에서 12h 이상 비율 **4.11%**를 보였고, `소각/농업활동 × 산림 내부형 × 악기상`도 **3.98%**로 높았다. `소각/농업활동 × 산림 경계/혼재형 × 악기상`은 12h 이상 비율 **3.36%**이지만 평균 피해면적이 매우 크게 튀어, 산림 경계부 소각은 빈번한 장시간 진화보다 초대형 피해 꼬리 위험으로 보는 것이 더 적절하다.

따라서 모델링 후보는 단순 `cause_group` 단독보다 `cause_group × landcover_context_type × weather_peak_day` 조합 또는 이를 축약한 `산림 내부/우세 + 악기상 + 원인그룹` 상호작용으로 구성하는 것이 낫다.

## 10. 평균 피해면적의 초대형 피해 꼬리 민감도

기존 결과에서 평균 피해면적이 높은 집단은 실제 전형적인 산불 규모가 큰 것인지, 소수 초대형 피해가 평균을 끌어올린 것인지 분리할 필요가 있다.
여기서는 악기상 조건으로 제한한 뒤 토지피복 맥락별 10ha 이상 비율, 중앙값, 90분위, 최대값, 상위 1건/5건의 전체 피해 기여도를 함께 확인한다.

```python
peak_weather_df = df[df["weather_peak_label"].eq("악기상")].copy()

def summarize_damage_tail(g):
    values = g["_damage_ha"].dropna().sort_values(ascending=False)
    total_damage = values.sum()
    return pd.Series({
        "전체 N": len(g),
        "10ha 이상 비율(%)": g["_large_damage_10ha"].mean() * 100,
        "평균 피해면적(ha)": values.mean(),
        "중앙 피해면적(ha)": values.median(),
        "90p 피해면적(ha)": values.quantile(0.9),
        "최대 피해면적(ha)": values.max(),
        "상위1건 피해비중(%)": values.iloc[0] / total_damage * 100 if len(values) and total_damage else np.nan,
        "상위5건 피해비중(%)": values.head(5).sum() / total_damage * 100 if len(values) and total_damage else np.nan,
    })

damage_tail_summary = (
    peak_weather_df.groupby("landcover_context_type", dropna=False)
    .apply(summarize_damage_tail, include_groups=False)
    .reset_index()
    .sort_values("평균 피해면적(ha)", ascending=False)
)
display(damage_tail_summary.round(4))
```

```python
tail_plot = damage_tail_summary.copy()
tail_plot["landcover_context_type"] = pd.Categorical(tail_plot["landcover_context_type"], context_order, ordered=True)
tail_plot = tail_plot.sort_values("landcover_context_type")

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
sns.barplot(data=tail_plot, x="landcover_context_type", y="10ha 이상 비율(%)", color="#7c6fb0", ax=axes[0])
axes[0].set_title("악기상 조건의 10ha 이상 피해 비율")
axes[0].set_xlabel("")
axes[0].tick_params(axis="x", rotation=35)

share_plot = tail_plot.melt(
    id_vars="landcover_context_type",
    value_vars=["상위1건 피해비중(%)", "상위5건 피해비중(%)"],
    var_name="꼬리 지표",
    value_name="전체 피해 중 비중(%)",
)
sns.barplot(data=share_plot, x="landcover_context_type", y="전체 피해 중 비중(%)", hue="꼬리 지표", ax=axes[1])
axes[1].set_title("평균 피해면적의 초대형 피해 의존도")
axes[1].set_xlabel("")
axes[1].tick_params(axis="x", rotation=35)
axes[1].legend(title="꼬리 지표")
plt.tight_layout()
fig.savefig(TMP_PLOT_DIR / "07_damage_tail_sensitivity.png", dpi=160, bbox_inches="tight")
plt.show()
```

### 10-1. 피해 꼬리 민감도 해석 및 플롯 이미지 분석

플롯을 직접 확인하면 10ha 이상 피해 비율은 `산림 우세형` **1.93%**, `산림 내부형` **1.27%**, `산림 경계/혼재형` **0.97%** 순으로 나타난다. 즉 대형 피해가 발생할 빈도 자체는 산림 내부성과 산림 우세성이 더 안정적인 신호다.

반면 평균 피해면적은 `산림 경계/혼재형`이 가장 커 보이지만, 악기상 조건에서 상위 1건이 전체 피해면적의 **61.35%**, 상위 5건이 **87.35%**를 차지한다. `농지/소각 인접형`도 상위 1건 비중이 **79.12%**로 높다. 따라서 평균 피해면적은 일반적인 피해 규모가 아니라 초대형 피해 꼬리의 존재를 나타내는 지표로 제한해서 해석해야 한다.

모델링에서는 평균 피해면적을 직접 목표로 쓰기보다 `10ha 이상 여부`, `12h 이상 진화 여부`, `log1p(피해면적)` 또는 분위수 기반 지표를 함께 검토하는 편이 안정적이다.

### 11-1. 최종 정리

이번 심화 분석에서 가장 중요한 보완점은 다음과 같다.

1. 기존 캐시의 `산림지역_ratio=0` 중 일부는 실제 비산림 산불이 아니라 원천 토지피복 매칭 없음이었다. 모델링 전 원천 매칭 없음 플래그를 반드시 분리해야 한다.
2. 토지피복도는 수종 세분류보다 발생점 피복, 산림 최단거리, 산림/도심/농지 경계 구조가 더 우선이다.
3. 발생일 악기상 조건을 결합하면 산림 내부형·산림 우세형·산림 경계/혼재형에서 장시간 진화 신호가 뚜렷해진다. 특히 산림 내부형은 표본 수와 12h 이상 비율이 모두 안정적이다.
4. 소각·농업활동 원인은 농지 내부보다 산림 경계/혼재형 및 산림 내부형과 결합할 때 더 위험하다. 따라서 원인 단독 피처보다 `발생원인 × 토지피복 맥락 × 악기상` 상호작용이 필요하다.
5. 평균 피해면적은 초대형 피해 한두 건에 매우 민감하다. 피해 규모 모델링은 평균 대신 `10ha 이상 여부`, `12h 이상 진화 여부`, `log1p(피해면적)` 등 꼬리 민감도를 낮춘 목표를 함께 검토해야 한다.
6. 산림 방향성은 단독 효과보다 풍향, 사면방향, 영동권 여부와 결합해 확산 위험을 설명하는 상호작용 후보로 남기는 것이 안전하다.

추가로 필요한 원천 데이터는 일반 도로망 또는 도로 중심선 데이터다. 현재 `nearest_road_dist_km`는 임도망도 기준이라 “차량 접근 가능한 일반 도로”를 대표하지 않는다.
